# Question 1 — List Operations

In this question, the individual digits of my roll number are extracted and each digit is multiplied by 10 to create a list.

The following list operations are performed:
- append()
- insert()
- remove()
- pop()
- sort()
- slicing
- list comprehension

```python
roll_no = "1024170445"

digits = [int(digit) for digit in roll_no]

L = [digit * 10 for digit in digits]

print("Original Digits: ", digits)
print("L: ", L)
```

## ii. Adding Elements to the List

The append() method adds an element to the end of the list.

The insert() method adds an element at a specific index.

```python
L.append(80)
print(L)

L.insert(10,100)
print(L)
```

iii. Remove two elements from L

```python
L.remove(20)

print(L)
L.pop(1)
print(L)
```

iv. Sort L in ascending order using sort(), then descending order using sort(reverse=True). Print both.

```python
L.sort()
print("Ascending Order:",L)

L.sort(reverse=True)
print("Descending Order:",L)
```

v. Use slicing to print the first three elements and the last three elements of L in a single line each.

```python
print(L[:3])
print(L[-3:])
```

vi. Use list comprehension to create a new list containing only the elements of L that are greater than the average of L.

```python
avg = sum(L) / len(L)
newList = [x for x in L if x > avg]
print(avg)
print(newList)
```

```python
even_numbers = [x for x in numbers if x % 2 == 0]
print("Total even numbers:",len(even_numbers))
print(even_numbers)
```

4. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))
```

5. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
freq = {x: numbers.count(x) for x in set(numbers)}

most_frequent = max(freq, key=freq.get)
count = freq[most_frequent]

print("\nMost frequent number:", most_frequent)
print("It occurs:", count, "time(s)")
```
