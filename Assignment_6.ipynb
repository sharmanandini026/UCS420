```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures (°C):", corrected_temperature)

fahrenheit = (9/5) * corrected_temperature + 32
print("Temperatures (°F):", fahrenheit)

above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", above_32)

count_above_32 = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count_above_32)
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])


total_steps = np.sum(steps)
print("Total steps:", total_steps)


mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)

print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

day_totals = np.sum(steps, axis=0)
print("Total steps for each day:", day_totals)

user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)


max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum value (user, day):", max_position)
```

```python

original = np.array([1, 2, 3, 4, 5, 6])
print("Original:", original)

subset = original[1:4]
print("Subset:", subset)

subset[0] = 999
print("After modifying subset:")
print("Original:", original)
print("Subset:", subset)

copied_array = original[1:4].copy()
copied_array[0] = 500
print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print("\n3 x 4 matrix:\n", matrix)


print("\nFirst row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[0:2, 1:3])

flattened = matrix.flatten()
raveled = matrix.ravel()
print("\nUsing flatten():", flattened)
print("Using ravel():", raveled)

raveled[0] = 1000
print("\nAfter modifying ravel():")
print("Raveled array:", raveled)
print("Original matrix:\n", matrix)

flattened[1] = 2000
print("\nAfter modifying flatten():")
print("Flattened array:", flattened)
print("Original matrix:\n", matrix)


print("\nShape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])


print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)


XT = X.T
print("\nX.T:\n", XT)

XTX = XT @ X
print("\nX.T @ X:\n", XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nInverse of X.T @ X:\n", XTX_inv)

beta = XTX_inv @ XT @ y
print("\nOLS coefficients (beta):", beta)


print("\nCoefficient interpretation:")
print("The coefficients correspond to sleep hours, activity level, and stress level respectively.")
print("Each coefficient indicates the change in predicted assistance score associated with a one-unit change in that feature, while the other features are held constant.")


new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("\nPredicted assistance score:", predicted_score)
```
